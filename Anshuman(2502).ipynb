in C, adding a new field to a struct requires changing the struct definition. WHat happens in Python when you add a new key to a dictionary while the program in running and why is that both useful and risky?
WHY IS IT USEFUL?
Flexibility with unpredictable or evolving data — great for JSON-like data, config files, API responses, or anything whose fields you don't know in advance.
Rapid prototyping — you can start using new fields without touching a class/struct definition anywhere.
Heterogeneous records — different dicts can carry different fields naturally (e.g., some user records have "phone", others don't), without needing subtypes or optional/nullable fields declared up front.
WHY IS IT RISKY?
No compile-time safety net. In C, if you typo a field name (p.naem), the compiler stops you cold. In Python, person["naem"] just raises a KeyError — at runtime, possibly in production, possibly only on some rare code path nobody tested.
No enforced schema. Nothing stops one part of your code from expecting person["age"] to exist while another part never sets it. You end up relying on convention and defensive code (.get("age", None)) instead of guarantees.
Harder to reason about and refactor. With a struct, you can see every field in one place. With dicts passed around and mutated in different functions, the "shape" of the data can drift silently — one function adds a key, another doesn't know it's there, a third assumes it always is.
Debugging cost. Bugs from missing/misnamed keys surface late (at access time) rather than early (at compile time), often far from where the mistake was introduced.

The C program declares that marks is a float.The python dictionary does not declare anything.Where does the type come from?
In C: the type is declared in advance, on the variable/field itself:The compiler reserves a fixed-size chunk of memory for a float, before any value is even assigned. If you later tried marks = "hello";, it would be a compile error — the type is locked in.

In Python: there's no variable/field declaration at all. Types belong to the values, not to the names or dictionary keys.
